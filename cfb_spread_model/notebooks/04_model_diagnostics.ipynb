# 04 - Model diagnostics

Exploratory companion to `scripts/train_models.py` and `scripts/evaluate_models.py`. Run both
first, then use this notebook for the precision-recall curve and calibration-by-bucket views
that don't belong in the non-interactive pipeline output.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent / "src"))

import pandas as pd

from cfb_spread_model.modeling import evaluation
from cfb_spread_model.utils.paths import OUTPUTS_MODEL_COMPARISON

oof_df = pd.read_csv(OUTPUTS_MODEL_COMPARISON / "oof_predictions.csv")
oof_df["model_name"].unique()
```

```python
model_name = "xgboost"
sub = oof_df[oof_df["model_name"] == model_name]
pr_table = evaluation.precision_recall_curve_table(sub["y_true"].to_numpy(), sub["y_score"].to_numpy())
pr_table.plot(x="recall", y="precision", kind="line", title=f"{model_name}: precision-recall curve (all OOF folds pooled)")
```

```python
evaluation.calibration_by_predicted_bucket(sub["y_true"].to_numpy(), sub["y_score"].to_numpy())
```
