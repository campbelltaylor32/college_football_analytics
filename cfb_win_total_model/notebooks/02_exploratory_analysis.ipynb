# 02 - Exploratory Analysis

Notebook companion to `scripts/run_eda.py`. Loads the cached modeling dataset (`data/processed/modeling_dataset.parquet`, built by `scripts/build_modeling_dataset.py`) and explores it interactively.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent / "src"))

import pandas as pd
import matplotlib.pyplot as plt

from cfb_win_total_model.dataset import NON_FEATURE_COLS
from cfb_win_total_model.utils.paths import DATA_PROCESSED_DIR

df = pd.read_parquet(DATA_PROCESSED_DIR / "modeling_dataset.parquet")
df.shape
```

```python
df["regular_season_wins"].describe()
```

```python
df["regular_season_wins"].hist(bins=range(0, 16))
plt.xlabel("Regular season wins")
plt.title("Distribution of regular-season wins")
plt.show()
```

```python
df.boxplot(column="regular_season_wins", by="season", rot=90, figsize=(10, 5))
plt.title("Wins by season")
plt.suptitle("")
plt.show()
```

```python
# Prior-year wins vs next-year wins -- the single strongest sanity-check scatter
plt.scatter(df["prior_season_wins"], df["regular_season_wins"], alpha=0.3)
plt.xlabel("prior_season_wins (t-1)")
plt.ylabel("regular_season_wins (t)")
plt.title("Prior-year wins vs next-year wins")
plt.show()
```

```python
feature_cols = [c for c in df.columns if c not in NON_FEATURE_COLS]
numeric_cols = df[feature_cols].select_dtypes(include="number").columns
corr = df[list(numeric_cols) + ["regular_season_wins"]].corr()["regular_season_wins"].drop("regular_season_wins")
corr.abs().sort_values(ascending=False).head(20)
```

```python
# Missingness by column
df.isna().mean().sort_values(ascending=False).head(20)
```
